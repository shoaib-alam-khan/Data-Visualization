# Prove Why We Visualize

### Dataset: datasaurus.csv

This task uses 13 datasets that have very similar summary statistics.

The main question is:

--> Are summary statistics alone enough to understand a dataset?

We will calculate the statistics first and then use scatter plots to see the actual shapes.

## 1. Import libraries


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Load the data


```python
data = pd.read_csv("data/datasaurus.csv")

data.head()
```

## 3. Check the dataset


```python
print("Rows:", len(data))
print("Columns:", list(data.columns))

data.info()
```

## 4. Calculate summary statistics



```python
summary = data.groupby("dataset").agg(
    mean_x=("x", "mean"),
    mean_y=("y", "mean"),
    std_x=("x", "std"),
    std_y=("y", "std")
)

correlation = data.groupby("dataset")[["x", "y"]].corr().iloc[0::2, -1]
correlation.index = correlation.index.droplevel(1)
summary["correlation"] = correlation

summary.round(2)
```

## 5. What would we conclude from the table?

If the table was the only thing we had, the datasets would look very similar.

We might conclude that the relationship between x and y is also very similar in every dataset. But this conclusion is incomplete because summary statistics do not show the actual shape of the data.

The next chart let us see the data.

## 6. Plot all 13 datasets

A scatter plot is suitable because we want to see the relationship between two numeric variables, x and y.

We use one small plot for each dataset so their shapes can be compared directly.

```python
names = data["dataset"].unique()

fig, axes = plt.subplots(4, 4, figsize=(14, 12))
axes = axes.flatten()

for i, name in enumerate(names):
    part = data[data["dataset"] == name]

    axes[i].scatter(part["x"], part["y"], s=12)
    axes[i].set_title(name)
    axes[i].set_xlabel("x")
    axes[i].set_ylabel("y")

for i in range(len(names), len(axes)):
    axes[i].axis("off")

plt.suptitle("Similar Statistics Can Hide Very Different Data Shapes", fontsize=16)
plt.tight_layout()
plt.show()
```

## 7. What did we actually find?

The scatter plots show that the datasets can have very different shapes even when their summary statistics are close.

Some datasets form clear shapes, while others have lines, curves or other patterns. This means that the mean, standard deviation and correlation are not enough to understand the complete structure of the data.

### Answer to the task question

Having 13 datasets and 142 points in each dataset does not make summary statistics a complete replacement for visualization. More observations can give more stable statistics, but the statistics can still hide important patterns in the data.

**Main:** always look at the data, not only its summary statistics.
