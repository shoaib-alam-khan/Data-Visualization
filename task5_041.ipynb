#  A Real Time Series

### Dataset: flights.csv

This dataset contains monthly airline passenger totals from 1949 to 1960.

The main challenge is creating one ordered time variable from year and mont.

## 1. Import libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Load the data

```python
data = pd.read_csv("data/flights.csv")

data.head()
```

## 3. Create an ordered time column



```python
data["date"] = pd.to_datetime(
    data["year"].astype(str) + "-" + data["month"],
    format="%Y-%B"
)

data = data.sort_values("date")

data.head()
```

## 4. Passengers over time

```python
plt.figure(figsize=(12, 6))

plt.plot(data["date"], data["passengers"])

plt.title("Airline Passenger Numbers Rose Strongly from 1949 to 1960")
plt.xlabel("Year")
plt.ylabel("Number of passengers")
plt.show()
```

### Why this chart?

A line chart is appropriate because the data has an ordered time axis.

The chart shows the long term trend, but it also contains a repeating seasonal pattern.

## 5. Isolate the seasonal cycle

```python
plt.figure(figsize=(12, 6))

for year in data["year"].unique():
    part = data[data["year"] == year]
    plt.plot(part["month"], part["passengers"], label=str(year))

plt.title("Passenger Seasonality Repeats Within Each Year")
plt.xlabel("Month")
plt.ylabel("Number of passengers")
plt.legend(title="Year", ncol=2)
plt.show()
```

### Why this chart?

This chart keeps month on the x-axis and gives each year its own line.

It makes the repeating within-year cycle easier to see than the first time-series chart.

## 6. Which month is consistently busiest?

```python
month_average = data.groupby("month")["passengers"].mean()
month_average = month_average.reindex([
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
])

plt.figure(figsize=(12, 6))

plt.bar(month_average.index, month_average.values)

plt.title("July Has the Highest Average Passenger Total")
plt.xlabel("Month")
plt.ylabel("Average passengers")
plt.xticks(rotation=45)
plt.show()
```

### Why this chart?

A bar chart is useful for comparing the average passenger total across the 12 months.

Because the months are being compared as categories, a bar chart makes the busiest month easy to see.

## 7. Is the bigger seasonal swing caused by growth?

The seasonal swing becomes larger in absolute passenger numbers over time.

To check whether the seasonality itself is growing, we can look at passengers as a percentage of each year's average. This removes much of the effect of the overall growth of the airline.

```python
data["year_average"] = data.groupby("year")["passengers"].transform("mean")
data["seasonal_index"] = data["passengers"] / data["year_average"]

plt.figure(figsize=(12, 6))

for year in data["year"].unique():
    part = data[data["year"] == year]
    plt.plot(part["month"], part["seasonal_index"], label=str(year))

plt.axhline(1, linestyle="--")
plt.title("Relative Seasonality Can Be Compared After Removing Overall Growth")
plt.xlabel("Month")
plt.ylabel("Passengers / yearly average")
plt.legend(title="Year", ncol=2)
plt.show()
```

### Answer to the question

The absolute seasonal swing can become larger simply because the total number of passengers is increasing.

A normalized seasonal-index chart, like the one above, helps separate the repeating seasonal pattern from overall growth.

### Report choice

For a report, the first time-series chart is useful for showing the long-term growth, while the seasonal chart is useful when the report specifically discusses seasonality. The month comparison chart is more focused on the busiest month and may not need to appear beside both of the other charts.
